```sql
SELECT
    pickup_date,
    total_trips,
    ROUND(total_revenue, 2) AS total_revenue,
    ROUND(average_fare, 2) AS average_fare,
    ROUND(average_trip_distance_miles, 2) AS average_trip_distance_miles,
    ROUND(average_trip_duration_minutes, 2) AS average_trip_duration_minutes
FROM taxicatalog.`taxi_project-schema`.daily_taxi_metrics
ORDER BY pickup_date;
```
